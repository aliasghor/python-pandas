````python
# Import the Python Pandas package, and set it as pd for the alias package name
import pandas as pd

# Import the Python NumPy package, and set it as np for the alias package name
import numpy as np

# Print the Python Pandas package version
print(pd.__version__)

# Print the Python NumPy package version
print(pd.__version__)
````

# DataFrame

A `DataFrame` is a 2-dimensional array that stores data of different types, such as integers, floating-point values, strings, lists, etc., in form of a table. To manually store data in a table, we'll need to use the `DataFrame` class. When we're using a Python dictionary of lists, the keys will be used as column headers, whereas the values will automatically correspond the column headers.

For instance, below is the example to manually store data in a table:

````python
data = {
    "Name": ["Braund, Mr. Owen Harris", "Allen, Mr. William Henry", "Bonnel, Miss. Elizabeth"],
    "Age": [22, 35, 58],
    "Sex": ["Male", "Male", "Female"]
}

df = pd.DataFrame(data)
df
````

# Series

A `Series` is a one-dimensional array that contains data of the same type. It also provides the ability to have labels for the indices, instead of normal integer indices.

````python
# Create Python Pandas Series that has a sequence of floating-point values
nums = pd.Series([-1, -0.5, 0, 0.5, 1], dtype=np.float16) # The -1 and 1 integer values are automatically casted into floating-point values
nums
````

````python
# Indexing using integer value
print(f"The first index of nums has a value of = {nums[0]}")
print(f"The last index of nums has a value of = {nums[4]}")
````

Please do note that in Pandas, we can't use the -1 integer index to obtain the last value in a Series.

````python
# Use labels for the Series indices
grades = pd.Series([60, 85, 30, 100, 50], index=["Ali", "Gerry", "Mogi", "Pace", "Balu"])
grades
````

````python
# Indexing using labels
print(f"The first index of grades have a value of: {grades['Ali']}")
````
