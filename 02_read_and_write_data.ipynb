# 2. Read And Write Data

```python
# Import the Pandas package, and set it as pd for the alias name
import pandas as pd
```

# Read Available Data Using `pd.read_csv()` function

The `pd.read_csv()` function is used to read a CSV (Comma Seperated Values) file into a Pandas `DataFrame` object.

```python
# Read the titanic csv file using the `pd.read_csv()` function
df = pd.read_csv("titanic.csv")

# When displaying a DataFrame, the first and last five rows will be shown by default
df
```

To display five rows of a `DataFrame`, we'll need to use the `.head()` method:

```python
# Display only five rows of a DataFrame
df.head()
```

We can use the `df.head()` method with a specified integer arugment to shows how many rows that want to be displayed:

```python
# Display only the first ten rows of a DataFrame
df.head(10)
```

We can also use the `df.tail()` to shows only the last rows of a `DataFrame`. It also has an argument to specify how many last rows of a `DataFrame` that want to be displayed.

```python
# Display the last five rows of a DataFrame
df.tail()
```

```python
# Display the last ten rows of a DataFrame
df.tail(10)
```

# Checking Column Data Types

To check the data types of each column, we'll need to use the `.dtypes` attribute.

```python
# Check each column data types
df.dtypes
```

> **Note📖**: In Pandas, `object` is a `str` (string) data type.

# Convert And Store A Specific File

There are several ways to convert and store data, one of them is by using the `to._excel()` function. It is used to store data as an excel file.

```python
# Storing the data as an excel file
df.to_excel("titanic.xlsx", sheet_name="passengers", index=False)
```

In the example above, the `sheet_name` is named passengers instead of the default name, which is Sheet1. By setting `index=False`, the row index labels aren't displayed in the columns.

```python
# Read the titanic.xlsx file
df_excel = pd.read_excel("titanic.xlsx")

# Display only the first five rows
df_excel.head()
```

# Show Technical Informations Using the `.info()` Method

The `.info()` method shows a technical information of a `DataFrame`. Below is the example of using the `.info()` method:

```python
df.info()
```

## Output Explanations:

- **<class 'pandas.core.frame.DataFrame'>**: The object data type.
- **RangeIndex: 891 entries, 0 to 890**: Show how many rows that were available.
- **Data columns (total 12 columns)**: There are twelve columns in a `DataFrame` object.
  - **#**: The total columns number (starts from 0 until 11).
  - **Column**: The columns name that were available.
  - **Non-Null Count**: Shows how many missing values that were available. For instance, the column **Cabin** has a total missing values for about 686 rows (891 - 204).
  - **Dtype**: Shows each data type of the columns `DataFrame`.
- **dtypes**: Show the available data types
- **memory usage**: The approximate amount of RAM used to hold the DataFrame.
